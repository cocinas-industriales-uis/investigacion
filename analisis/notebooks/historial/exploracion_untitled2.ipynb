# Podemos deducir
De acuerdo con la Clasificación Industrial Internacional Uniforme adaptada para Colombia, las cocinas industriales no constituyen una clase económica independiente. Su clasificación depende del servicio de alimentación desarrollado. Para esta investigación se selecciona principalmente la clase CIIU 5629, correspondiente a “Actividades de otros servicios de comidas”, debido a que comprende la operación de comedores y servicios de alimentación en fábricas, oficinas, hospitales, instituciones educativas y otras instalaciones. Este contexto se relaciona directamente con el proyecto, dado que comprende cocinas de operación continua, preparación de alimentos a mayor escala y utilización de equipos que requieren supervisión de variables como gas, temperatura y presencia de llama.

```python
import pandas as pd

# Create a dictionary with your data
data = {
    'Año': [ 2023, 2024,2025],
    'Trabajadores': [78.549, 221.463,263.602],
    'Accidentes': [5.410 , 14.674,16.062],
    'Tasa por 100 trabajadores': [6.88,6.63,6.09],
    'Fuente': [ 'Fasecolda', 'Fasecolda','Fasecolda']



}

# Create a DataFrame (table) from the dictionary
df = pd.DataFrame(data)

# Display the DataFrame
display(df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Convert 'Trabajadores' and 'Accidentes' columns to numeric, handling the thousand separators
df['Trabajadores'] = df['Trabajadores'].astype(str).str.replace('.', '').astype(int)
df['Accidentes'] = df['Accidentes'].astype(str).str.replace('.', '').astype(int)

# Set up the matplotlib figure and a secondary axis
fig, ax1 = plt.subplots(figsize=(12, 6))

# Plot 'Trabajadores' and 'Accidentes' as bar charts on the primary y-axis (ax1)
width = 0.35 # Width of the bars
ax1.bar(df['Año'] - width/2, df['Trabajadores'], width, label='Trabajadores', color='skyblue')
ax1.bar(df['Año'] + width/2, df['Accidentes'], width, label='Accidentes', color='lightcoral')

ax1.set_xlabel('Año')
ax1.set_ylabel('Cantidad (Trabajadores / Accidentes)', color='black')
ax1.tick_params(axis='y', labelcolor='black')
ax1.set_xticks(df['Año'])

# Create a secondary y-axis for 'Tasa por 100 trabajadores'
ax2 = ax1.twinx()
ax2.plot(df['Año'], df['Tasa por 100 trabajadores'], color='green', marker='o', linestyle='-', linewidth=2, label='Tasa por 100 trabajadores')
ax2.set_ylabel('Tasa por 100 trabajadores', color='green')
ax2.tick_params(axis='y', labelcolor='green')

# Add title and legend
plt.title('Análisis de Trabajadores, Accidentes y Tasa de Accidentalidad por Año')

# Combine legends from both axes
h1, l1 = ax1.get_legend_handles_labels()
h2, l2 = ax2.get_legend_handles_labels()
ax1.legend(h1+h2, l1+l2, loc='upper left')

plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout() # Adjust layout to prevent overlapping elements
plt.show()
```

Entre 2023 y 2025 aumentaron tanto los trabajadores afiliados como el número absoluto de accidentes laborales. Sin embargo, la tasa de accidentalidad disminuyó de 6,88 a 6,09 accidentes por cada 100 trabajadores. Esto indica que los accidentes crecieron en cantidad, pero disminuyeron proporcionalmente respecto a la población trabajadora. Los resultados corresponden al sector de alojamiento y servicios de comida, por lo que no representan exclusivamente accidentes ocurridos en cocinas industriales.

Entre 2010 y 2014, los departamentos de bomberos de Estados Unidos atendieron un promedio estimado de 7.410 incendios estructurales anuales en establecimientos de comidas y bebidas. Los equipos de cocina estuvieron involucrados en el 61 % de estos incendios y el 59 % comenzó específicamente en la cocina o zona de preparación.

Entre los equipos involucrados se destacaron las freidoras, presentes en el 21 % de los incendios, y las estufas o superficies de cocción, con el 14 %. Además, la falta de limpieza fue identificada como un factor contribuyente en el 22 % de los casos.

Estos resultados muestran que una parte importante del riesgo se concentra en los equipos que producen calor y en las áreas donde se preparan los alimentos. Esto respalda la utilización del sensor de temperatura DS18B20 y del sensor de llama KY-026, junto con los sistemas de ventilación y alarma del prototipo, para identificar tempranamente condiciones peligrosas.

Las investigaciones revisadas respaldan el uso de sistemas IoT para monitorear continuamente variables relacionadas con fugas de gas, temperaturas elevadas e incendios. Estos sistemas pueden procesar la información de los sensores, generar alarmas y activar respuestas automáticas como la ventilación.

Mensch et al. evaluaron 16 sensores durante 60 experimentos realizados en una cocina simulada. Los resultados mostraron que la combinación de diferentes variables puede ayudar a distinguir una situación peligrosa de una actividad normal de cocción. Este hallazgo respalda la decisión de integrar en el prototipo sensores de temperatura, gas y llama, en lugar de depender de una sola medición. [Investigación publicada por NIST](https://www.nist.gov/publications/sensors-and-machine-learning-models-prevent-cooktop-ignition-and-ignore-normal-cooking).

Jena et al. desarrollaron un prototipo para detectar fugas de GLP que, al superar un nivel determinado, activaba una alarma, un sistema de comunicación y un ventilador extractor. Esto presenta una arquitectura semejante a la propuesta en este proyecto: detección, procesamiento, alerta y actuación automática. .[Artículo en Materials Today: Proceedings](https://www.sciencedirect.com/science/article/abs/pii/S2214785322070225?via%3Dihub)

Kodali et al. integraron sensores MQ-6 para GLP y utilizaron un ESP32 para la comunicación del sistema. Esta investigación constituye un antecedente directo para la selección del sensor MQ-6 y del microcontrolador ESP32. Artículo publicado por IEEE[ Artículo publicado por IEEE](https://ieeexplore.ieee.org/document/8777463).

Maltezos et al. combinaron sensores de temperatura, humo, llama y gases con un ESP32 y procesamiento local. El sistema publicó mediciones y alertas en tiempo real, lo que respalda el uso de procesamiento cercano a los sensores para reducir la dependencia de una conexión externa. Artículo publicado en Information[Artículo publicado en Information](https://www.mdpi.com/2078-2489/13/4/164).

La literatura científica demuestra que la integración de sensores, microcontroladores, comunicación IoT y actuadores es una alternativa técnicamente viable para monitorear riesgos relacionados con gas, temperatura y fuego. Los antecedentes encontrados respaldan la arquitectura general del prototipo propuesto, especialmente el uso del ESP32, el sensor MQ-6, el monitoreo térmico, la detección de llama y la activación de alarmas y ventilación.

Sin embargo, estos estudios no garantizan automáticamente la efectividad del prototipo, porque utilizaron componentes, condiciones y escenarios diferentes. Por tanto, será necesario validar experimentalmente el tiempo de detección, la repetibilidad, las falsas alarmas, los eventos no detectados y el comportamiento del sistema cuando falle la conexión.

En conclusión, la solución IoT está respaldada científicamente como concepto, pero su desempeño específico deberá comprobarse mediante pruebas controladas antes de considerarla un sistema de seguridad funcional.
